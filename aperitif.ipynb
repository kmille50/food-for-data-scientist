```python
# CORRECTION 

from typing import List
import pandas as pd


def clean_data(data: List[dict]) -> float:
    df = pd.DataFrame(data)

    df["age"] = pd.to_numeric(df["age"], errors="coerce")
    df = df.dropna(subset=["age"])

    return round(df["age"].mean(), 2)
```
